# Analyse croisée du lieu de décès – Dr Death

Ce notebook présente une analyse du lieu de décès des victimes de Shipman, en fonction du sexe, de la tranche d’âge et de l’âge réel.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Chargement des données
victims = pd.read_csv("shipman-confirmed-victims.csv")
```

##  Lieu de décès selon le sexe

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=victims, x='PlaceofDeath', hue='gender2', order=victims['PlaceofDeath'].value_counts().index)
plt.title("Lieu de décès selon le sexe")
plt.xlabel("Lieu de décès")
plt.ylabel("Nombre de victimes")
plt.legend(title="Sexe", labels=["Homme", "Femme"])
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Lieu de décès selon la tranche d’âge

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=victims, x='PlaceofDeath', hue='ageBracket', order=victims['PlaceofDeath'].value_counts().index)
plt.title(" Lieu de décès selon la tranche d’âge")
plt.xlabel("Lieu de décès")
plt.ylabel("Nombre de victimes")
plt.legend(title="Tranche d’âge", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Âge des victimes selon le lieu de décès

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=victims, x='PlaceofDeath', y='Age', order=victims['PlaceofDeath'].value_counts().index)
plt.title("Distribution de l’âge selon le lieu de décès")
plt.xlabel("Lieu de décès")
plt.ylabel("Âge")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
