# Modèle de classification – Prédire si le décès concerne Shipman

Ceci un modèle de machine learning supervisé (Random Forest) pour prédire si un décès a été causé par Harold Shipman, en se basant sur des données telles que l'heure, le sexe, l'âge, et le lieu du décès.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Chargement des données horaires avec cible
df = pd.read_csv('data/shipman-times-comparison.csv')

# Vérification
df.head()
```

```python
# Nettoyage des données : encoder les colonnes texte si nécessaire
if df['Comparison'].dtype == 'object':
    df['Comparison'] = LabelEncoder().fit_transform(df['Comparison'])

# Préparation des features et de la cible
X = df[['Hour', 'Comparison']]
y = df['Shipman']
```

```python
# Split train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X = df[['Hour', 'Comparison']]
y = df['Shipman'].astype(int)  # 🔧 conversion explicite
```

```python
# Entraînement du modèle (fit)
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Vérification que le modèle est bien entraîné
print("Modèle entraîné avec", len(model.estimators_), "arbres.")

# Prédiction sur les données test
y_pred = model.predict(X_test)

# Évaluation
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Importance des variables
importances = pd.Series(model.feature_importances_, index=X.columns)
importances.sort_values().plot(kind='barh', color='skyblue')
plt.title("Importance des variables pour la prédiction")
plt.show()
```

```python
X = df[['Hour', 'Comparison']]
y = df['Shipman']

# Créer un label binaire Shipman / Non-Shipman
y_binary = (y == 1).astype(int)

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y_binary, test_size=0.3, random_state=42)

from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
```
